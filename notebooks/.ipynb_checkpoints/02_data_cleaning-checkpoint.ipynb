```python
import pandas as pd
import numpy as np
import sqlite3
import json
from pathlib import Path
from sqlalchemy import create_engine, text

RAW     = Path("../data/raw")
PROCESSED = Path("../data/processed")
DB_DIR  = Path("../data/db")

PROCESSED.mkdir(parents=True, exist_ok=True)
DB_DIR.mkdir(parents=True, exist_ok=True)

print("Libraries imported")
print(f"RAW folder      : {RAW}")
print(f"PROCESSED folder: {PROCESSED}")
print(f"DB folder       : {DB_DIR}")
```

```python
df_nav = pd.read_csv(RAW / "02_nav_history.csv")

print("BEFORE cleaning:")
print(f"Shape   : {df_nav.shape}")
print(f"Columns : {df_nav.columns.tolist()}")
print(f"Nulls   :\n{df_nav.isnull().sum()}")
print(f"\nhead(3):\n{df_nav.head(3)}")

df_nav["date"] = pd.to_datetime(df_nav["date"], errors="coerce")

df_nav = df_nav.sort_values(["amfi_code", "date"]).reset_index(drop=True)

df_nav["nav"] = pd.to_numeric(df_nav["nav"], errors="coerce")
df_nav["nav"] = df_nav.groupby("amfi_code")["nav"].ffill()

before = len(df_nav)
df_nav = df_nav.drop_duplicates(subset=["amfi_code", "date"])
print(f"\nDuplicates removed: {before - len(df_nav)}")

invalid_nav = df_nav[df_nav["nav"] <= 0]
print(f"Invalid NAV rows (<=0): {len(invalid_nav)}")
df_nav = df_nav[df_nav["nav"] > 0]

df_nav["daily_return_pct"] = df_nav.groupby("amfi_code")["nav"].pct_change() * 100

print(f"\nAFTER cleaning:")
print(f" Shape : {df_nav.shape}")
print(f" Nulls :\n{df_nav.isnull().sum()}")
print(f"\nhead(3):\n{df_nav.head(3)}")

df_nav.to_csv(PROCESSED / "clean_nav_history.csv", index=False)
print(f"\nSaved → data/processed/clean_nav_history.csv")
```

```python
df_txn = pd.read_csv(RAW / "08_investor_transactions.csv")

print("BEFORE cleaning:")
print(f"Shape   : {df_txn.shape}")
print(f"Columns : {df_txn.columns.tolist()}")
print(f"Nulls   :\n{df_txn.isnull().sum()}")

df_txn["transaction_date"] = pd.to_datetime(
    df_txn["transaction_date"], errors="coerce"
)

df_txn["transaction_type"] = (
    df_txn["transaction_type"]
    .str.strip()
    .str.title()
    .replace({
        "Sip": "SIP",
        "Lumpsum": "Lumpsum",
        "Redemption": "Redemption"
    })
)
print(f"\nTransaction types: {df_txn['transaction_type'].unique()}")

before = len(df_txn)
df_txn = df_txn[df_txn["amount_inr"] > 0]
print(f"Invalid amount rows removed: {before - len(df_txn)}")

print(f"\nKYC status values: {df_txn['kyc_status'].unique()}")
valid_kyc = ["Verified", "Pending"]
df_txn = df_txn[df_txn["kyc_status"].isin(valid_kyc)]

df_txn = df_txn.drop_duplicates()

print(f"\nAFTER cleaning:")
print(f"  Shape : {df_txn.shape}")
print(f"\nhead(3):\n{df_txn.head(3)}")

df_txn.to_csv(PROCESSED / "clean_investor_transactions.csv", index=False)
print(f"\nSaved → data/processed/clean_investor_transactions.csv")
```

```python
df_perf = pd.read_csv(RAW / "07_scheme_performance.csv")

print("BEFORE cleaning:")
print(f"  Shape   : {df_perf.shape}")
print(f"  Columns : {df_perf.columns.tolist()}")
print(f"  dtypes  :\n{df_perf.dtypes}")

numeric_cols = [
    "return_1yr_pct", "return_3yr_pct", "return_5yr_pct",
    "sharpe_ratio", "sortino_ratio", "alpha", "beta",
    "std_dev_ann_pct", "max_drawdown_pct"
]
for col in numeric_cols:
    if col in df_perf.columns:
        df_perf[col] = pd.to_numeric(df_perf[col], errors="coerce")

if "sharpe_ratio" in df_perf.columns:
    neg_sharpe = df_perf[df_perf["sharpe_ratio"] < 0]
    print(f"\nFunds with negative Sharpe: {len(neg_sharpe)}")
    if len(neg_sharpe) > 0:
        print(neg_sharpe[["amfi_code", "sharpe_ratio"]])

if "expense_ratio_pct" in df_perf.columns:
    df_perf["expense_ratio_pct"] = pd.to_numeric(
        df_perf["expense_ratio_pct"], errors="coerce"
    )
    out_of_range = df_perf[
        (df_perf["expense_ratio_pct"] < 0.1) |
        (df_perf["expense_ratio_pct"] > 2.5)
    ]
    print(f"Expense ratio out of range: {len(out_of_range)}")
    
df_perf = df_perf.drop_duplicates()

print(f"\nAFTER cleaning:")
print(f"  Shape : {df_perf.shape}")
print(f"  Nulls :\n{df_perf.isnull().sum()}")
print(f"\nhead(3):\n{df_perf.head(3)}")

df_perf.to_csv(PROCESSED / "clean_scheme_performance.csv", index=False)
print(f"\nSaved → data/processed/clean_scheme_performance.csv")
```

```python
OTHER_FILES = {
    "01_fund_master.csv":         "clean_fund_master.csv",
    "03_aum_by_fund_house.csv":   "clean_aum_by_fund_house.csv",
    "04_monthly_sip_inflows.csv": "clean_monthly_sip_inflows.csv",
    "05_category_inflows.csv":    "clean_category_inflows.csv",
    "06_industry_folio_count.csv":"clean_industry_folio_count.csv",
    "09_portfolio_holdings.csv":  "clean_portfolio_holdings.csv",
    "10_benchmark_indices.csv":   "clean_benchmark_indices.csv",
}

for raw_file, clean_file in OTHER_FILES.items():
    df = pd.read_csv(RAW / raw_file, low_memory=False)
    
    for col in df.columns:
        if "date" in col.lower() or "month" in col.lower():
            df[col] = pd.to_datetime(df[col], errors="coerce")

    df = df.drop_duplicates()

    for col in df.select_dtypes(include=np.number).columns:
        df[col] = df[col].fillna(df[col].median())

    df.to_csv(PROCESSED / clean_file, index=False)
    print(f"{raw_file:<35} → {clean_file}  shape={df.shape}")

print("\nAll 7 files cleaned and saved!")
```

```python
DB_PATH = DB_DIR / "bluestock_mf.db"
engine  = create_engine(f"sqlite:///{DB_PATH}")

schema_sql = """
-- dim_fund
CREATE TABLE IF NOT EXISTS dim_fund (
    amfi_code       TEXT PRIMARY KEY,
    fund_house      TEXT,
    scheme_name     TEXT,
    category        TEXT,
    sub_category    TEXT,
    plan            TEXT,
    benchmark       TEXT,
    expense_ratio_pct REAL,
    risk_category   TEXT,
    fund_manager    TEXT
);

-- dim_date
CREATE TABLE IF NOT EXISTS dim_date (
    date_id     TEXT PRIMARY KEY,
    year        INTEGER,
    month       INTEGER,
    quarter     INTEGER,
    month_name  TEXT,
    is_weekday  INTEGER
);

-- fact_nav
CREATE TABLE IF NOT EXISTS fact_nav (
    nav_id          INTEGER PRIMARY KEY AUTOINCREMENT,
    amfi_code       TEXT REFERENCES dim_fund(amfi_code),
    date            TEXT,
    nav             REAL,
    daily_return_pct REAL
);

-- fact_transactions
CREATE TABLE IF NOT EXISTS fact_transactions (
    txn_id          INTEGER PRIMARY KEY AUTOINCREMENT,
    investor_id     TEXT,
    amfi_code       TEXT REFERENCES dim_fund(amfi_code),
    transaction_date TEXT,
    transaction_type TEXT,
    amount_inr      REAL,
    state           TEXT,
    city_tier       TEXT,
    age_group       TEXT,
    gender          TEXT,
    kyc_status      TEXT
);

-- fact_performance
CREATE TABLE IF NOT EXISTS fact_performance (
    perf_id         INTEGER PRIMARY KEY AUTOINCREMENT,
    amfi_code       TEXT REFERENCES dim_fund(amfi_code),
    return_1yr_pct  REAL,
    return_3yr_pct  REAL,
    return_5yr_pct  REAL,
    sharpe_ratio    REAL,
    sortino_ratio   REAL,
    alpha           REAL,
    beta            REAL,
    max_drawdown_pct REAL
);

-- fact_aum
CREATE TABLE IF NOT EXISTS fact_aum (
    aum_id      INTEGER PRIMARY KEY AUTOINCREMENT,
    fund_house  TEXT,
    quarter     TEXT,
    aum_crore   REAL
);
"""
with engine.connect() as conn:
    for statement in schema_sql.strip().split(";"):
        stmt = statement.strip()
        if stmt:
            conn.execute(text(stmt))
    conn.commit()

print("SQLite database created!")
print(f"Location: {DB_PATH}")

schema_path = Path("../sql/schema.sql")
schema_path.parent.mkdir(exist_ok=True)
schema_path.write_text(schema_sql)
print(f"schema.sql saved → sql/schema.sql")
```

```python
load_map = {
    "clean_fund_master.csv":           "dim_fund",
    "clean_nav_history.csv":           "fact_nav",
    "clean_investor_transactions.csv": "fact_transactions",
    "clean_scheme_performance.csv":    "fact_performance",
    "clean_aum_by_fund_house.csv":     "fact_aum",
}

for csv_file, table_name in load_map.items():
    path = PROCESSED / csv_file
    if not path.exists():
        print(f"Skipping {csv_file} — file not found")
        continue

    df = pd.read_csv(path, low_memory=False)
    df.to_sql(table_name, engine, if_exists="replace", index=False)
    print(f"Loaded {csv_file:<45} → {table_name:<25} ({len(df):,} rows)")

print("\nAll tables loaded into bluestock_mf.db!")

print("\n── Row counts in database ───────────────────")
with engine.connect() as conn:
    for table in ["dim_fund", "fact_nav", "fact_transactions",
                  "fact_performance", "fact_aum"]:
        try:
            count = conn.execute(text(f"SELECT COUNT(*) FROM {table}")).scalar()
            print(f"  {table:<25} : {count:,} rows")
        except:
            print(f"  {table:<25} : table not found")
```

```python
queries = {
    "Q1 — Top 5 funds by expense ratio": """
        SELECT scheme_name, fund_house, expense_ratio_pct
        FROM dim_fund
        ORDER BY expense_ratio_pct DESC
        LIMIT 5
    """,
    "Q2 — Funds with expense_ratio < 1%": """
        SELECT scheme_name, fund_house, expense_ratio_pct
        FROM dim_fund
        WHERE expense_ratio_pct < 1.0
        ORDER BY expense_ratio_pct
    """,
    "Q3 — Average NAV per month": """
        SELECT 
            amfi_code,
            strftime('%Y-%m', date) AS month,
            ROUND(AVG(nav), 4) AS avg_nav
        FROM fact_nav
        GROUP BY amfi_code, month
        ORDER BY month DESC
        LIMIT 20
    """,
    "Q4 — Top 5 funds by Sharpe ratio": """
        SELECT f.scheme_name, f.fund_house, p.sharpe_ratio
        FROM fact_performance p
        JOIN dim_fund f ON p.amfi_code = f.amfi_code
        ORDER BY p.sharpe_ratio DESC
        LIMIT 5
    """,
    "Q5 — Transactions by state": """
        SELECT state,
               COUNT(*) AS total_txns,
               ROUND(SUM(amount_inr), 2) AS total_amount
        FROM fact_transactions
        GROUP BY state
        ORDER BY total_amount DESC
    """,
    "Q6 — SIP vs Lumpsum vs Redemption split": """
        SELECT transaction_type,
               COUNT(*) AS count,
               ROUND(SUM(amount_inr), 2) AS total_amount
        FROM fact_transactions
        GROUP BY transaction_type
    """,
    "Q7 — Funds with highest 3yr returns": """
        SELECT f.scheme_name, f.fund_house, p.return_3yr_pct
        FROM fact_performance p
        JOIN dim_fund f ON p.amfi_code = f.amfi_code
        ORDER BY p.return_3yr_pct DESC
        LIMIT 5
    """,
    "Q8 — Average transaction by age group": """
        SELECT age_group,
               COUNT(*) AS txn_count,
               ROUND(AVG(amount_inr), 2) AS avg_amount
        FROM fact_transactions
        GROUP BY age_group
        ORDER BY avg_amount DESC
    """,
    "Q9 — Funds by risk category": """
        SELECT risk_category,
               COUNT(*) AS fund_count
        FROM dim_fund
        GROUP BY risk_category
        ORDER BY fund_count DESC
    """,
    "Q10 — Funds with negative max drawdown < -20%": """
        SELECT f.scheme_name, p.max_drawdown_pct, p.sharpe_ratio
        FROM fact_performance p
        JOIN dim_fund f ON p.amfi_code = f.amfi_code
        WHERE p.max_drawdown_pct < -20
        ORDER BY p.max_drawdown_pct
    """,
}

queries_output = []
with engine.connect() as conn:
    for title, sql in queries.items():
        print(f"\n{'═'*55}")
        print(f"  {title}")
        print('═'*55)
        try:
            result = pd.read_sql(sql.strip(), conn)
            print(result.to_string(index=False))
            queries_output.append(f"-- {title}\n{sql.strip()}\n")
        except Exception as e:
            print(f"Error: {e}")

queries_sql_path = Path("../sql/queries.sql")
queries_sql_path.write_text("\n\n".join(queries_output))
print(f"\n queries.sql saved → sql/queries.sql")
```

```python
data_dict = """
Data Dictionary — Bluestock MF Capstone

01_fund_master.csv / dim_fund
| Column | Type | Description |
|---|---|---|
| amfi_code | TEXT | Unique AMFI scheme code |
| fund_house | TEXT | AMC name |
| scheme_name | TEXT | Full scheme name |
| category | TEXT | Equity / Debt / Hybrid |
| sub_category | TEXT | Large Cap / Mid Cap etc. |
| expense_ratio_pct | REAL | Annual expense ratio % |
| risk_category | TEXT | Low / Moderate / High |
| fund_manager | TEXT | Primary fund manager |

02_nav_history.csv / fact_nav
| Column | Type | Description |
|---|---|---|
| amfi_code | TEXT | Foreign key to dim_fund |
| date | DATE | NAV date (business days) |
| nav | REAL | NAV value in Rs. |
| daily_return_pct | REAL | Daily return % |

08_investor_transactions.csv / fact_transactions
| Column | Type | Description |
|---|---|---|
| investor_id | TEXT | Unique investor ID |
| transaction_date | DATE | Date of transaction |
| amfi_code | TEXT | Fund invested in |
| transaction_type | TEXT | SIP / Lumpsum / Redemption |
| amount_inr | REAL | Amount in Rs. |
| state | TEXT | Investor state |
| city_tier | TEXT | T30 or B30 |
| age_group | TEXT | 18-25 / 26-35 etc. |
| kyc_status | TEXT | Verified / Pending |

07_scheme_performance.csv / fact_performance
| Column | Type | Description |
|---|---|---|
| amfi_code | TEXT | Foreign key to dim_fund |
| return_1yr_pct | REAL | 1 year return % |
| return_3yr_pct | REAL | 3 year CAGR % |
| return_5yr_pct | REAL | 5 year CAGR % |
| sharpe_ratio | REAL | Risk adjusted return |
| alpha | REAL | Excess return vs benchmark |
| beta | REAL | Market sensitivity |
| max_drawdown_pct | REAL | Worst peak to trough % |
"""

dict_path = Path("../data_dictionary.md")
dict_path.write_text(data_dict)
print("data_dictionary.md saved!")
print(data_dict)
```
