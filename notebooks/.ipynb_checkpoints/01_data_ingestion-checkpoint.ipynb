```python
import pandas as pd
import numpy as np
from pathlib import Path

RAW = Path("../data/raw")
print("Libraries loaded!")
print("RAW folder exists:", RAW.exists())
```

```python
CSV_FILES = {
    "fund_master":        "01_fund_master.csv",
    "nav_history":        "02_nav_history.csv",
    "aum_by_fund_house":  "03_aum_by_fund_house.csv",
    "monthly_sip":        "04_monthly_sip_inflows.csv",
    "category_inflows":   "05_category_inflows.csv",
    "folio_count":        "06_industry_folio_count.csv",
    "scheme_performance": "07_scheme_performance.csv",
    "transactions":       "08_investor_transactions.csv",
    "portfolio_holdings": "09_portfolio_holdings.csv",
    "benchmark_indices":  "10_benchmark_indices.csv",
}

dataframes = {}
for name, filename in CSV_FILES.items():
    path = RAW / filename
    if path.exists():
        df = pd.read_csv(path, low_memory=False)
        dataframes[name] = df
        print(f"\n{'='*55}")
        print(f"Dataset : {name}")
        print(f"Shape   : {df.shape}")
        print(f"dtypes  :\n{df.dtypes}")
        print(f"head(3) :\n{df.head(3)}")
    else:
        print(f"NOT FOUND: {filename}")
```

```python
df_fund = dataframes.get("fund_master")
if df_fund is not None:
    print("Unique Fund Houses:")
    print(df_fund["fund_house"].unique())
    print("\nUnique Categories:")
    print(df_fund["category"].unique())
    print("\nUnique Risk Grades:")
    if "risk_category" in df_fund.columns:
        print(df_fund["risk_category"].unique())
```

```python
df_nav = dataframes.get("nav_history")
if df_fund is not None and df_nav is not None:
    fm_codes  = set(df_fund["amfi_code"].dropna().astype(str))
    nav_codes = set(df_nav["amfi_code"].dropna().astype(str))
    missing   = fm_codes - nav_codes
    coverage  = (len(fm_codes) - len(missing)) / len(fm_codes) * 100
    print(f"fund_master codes  : {len(fm_codes)}")
    print(f"nav_history codes  : {len(nav_codes)}")
    print(f"Coverage           : {coverage:.1f}%")
    print(f"Missing in nav     : {len(missing)}")
    print("\nData Quality Status: PASS" if coverage >= 95 else "WARN")
```

```python
print("Day 1 Data Ingestion Complete!")
print(f"Datasets loaded: {len(dataframes)}/10")
for name, df in dataframes.items():
    print(f"  {name:<25} : {df.shape}")
```
